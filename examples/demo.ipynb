## Comparing Models

Use `compare` to evaluate multiple models side-by-side:

```python
bench2 = Benchmark(tasks=['paper_comprehension', 'idea_generation'])
r1 = bench2.run(model='mock-a')
r2 = bench2.run(model='mock-b')

print(f"{'Task':30s} {'mock-a':>10s} {'mock-b':>10s}")
print('-' * 52)
tasks = list(r1.results)
for i in range(len(tasks)):
    t = tasks[i].task_name
    s1 = r1.results[i].score
    s2 = r2.results[i].score
    print(f"{t:30s} {s1:>10.2f} {s2:>10.2f}")
print(f"{'AVERAGE':30s} {sum(r.score for r in r1.results)/len(r1.results):>10.2f} {sum(r.score for r in r2.results)/len(r2.results):>10.2f}")
```
